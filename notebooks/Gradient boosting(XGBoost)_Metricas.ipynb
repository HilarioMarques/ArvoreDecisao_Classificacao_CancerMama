As **árvores de decisão** são algoritmos fundamentais em aprendizado de máquina, conhecidos por sua simplicidade e facilidade de interpretação. Elas funcionam dividindo os dados em subconjuntos baseados em condições sobre os atributos, criando uma estrutura em forma de árvore que leva a decisões finais. No entanto, árvores individuais podem não ser suficientemente precisas ou podem sofrer de **overfitting**. Para superar essas limitações, técnicas de ensemble como o **Gradient Boosting** são utilizadas para construir modelos mais robustos.

## Revisão: Fundamentos das Árvores de Decisão

### Como Funcionam as Árvores de Decisão?

- **Divisão Recursiva:** As árvores de decisão particionam os dados em subconjuntos menores através de condições baseadas nos atributos.
- **Estrutura em Árvore:**
  - **Nós Internos:** Representam testes ou condições sobre os atributos.
  - **Ramos:** Correspondem aos resultados dos testes, direcionando para o próximo nó.
  - **Nós Folha:** Indicam a decisão final ou previsão.
- **Critérios de Divisão:** Utilizam métricas como **Índice Gini** e **Entropia** para selecionar os melhores atributos para dividir os dados.

### Limitações das Árvores de Decisão

- **Overfitting:** Árvores muito profundas podem se ajustar excessivamente aos dados de treinamento, capturando ruídos em vez de padrões reais.
- **Instabilidade:** Pequenas variações nos dados podem levar a estruturas de árvore completamente diferentes.
- **Desempenho Subótimo:** Podem não alcançar a melhor performance possível em tarefas complexas.


# Introdução ao Gradient Boosting em Ciência de Dados

## Introdução ao Gradient Boosting

### O Que é o Gradient Boosting?

O **Gradient Boosting** é um método de ensemble que constrói modelos preditivos robustos ao combinar múltiplos modelos fracos, geralmente árvores de decisão simples, de forma sequencial. Cada novo modelo é treinado para corrigir os erros cometidos pelos modelos anteriores, aprimorando progressivamente a performance geral. O termo "gradient" refere-se ao uso do gradiente da função de perda para otimizar o modelo.

### Por Que Usar o Gradient Boosting?

- **Alta Precisão:** Pode alcançar resultados superiores em comparação com modelos individuais.
- **Flexibilidade:** Suporta diversas funções de perda e pode ser aplicado a problemas de classificação e regressão.
- **Controlável:** Oferece diversos hiperparâmetros para ajustar o modelo às necessidades específicas do problema.

## Como o Gradient Boosting Funciona?

### 1. Modelo Inicial

- **Previsão Básica:** O processo começa com uma previsão inicial simples. Por exemplo, na regressão, pode ser a média dos valores alvo.

### 2. Cálculo dos Resíduos

- **Erros de Previsão:** Calcula-se a diferença entre as previsões atuais e os valores reais (resíduos).
- **Objetivo:** Os resíduos representam o que o modelo precisa aprender a prever nas próximas iterações.

### 3. Treinamento de um Novo Modelo nos Resíduos

- **Nova Árvore:** Treina-se uma nova árvore de decisão para prever os resíduos.
- **Taxa de Aprendizado ($ \eta $):** Controla o quanto cada árvore nova contribui para o modelo final. Valores menores podem levar a um modelo mais robusto.

### 4. Atualização das Previsões

- **Combinação das Previsões:** As previsões do novo modelo são combinadas com as previsões anteriores para melhorar a precisão.
- **Fórmula Geral:**
  $$
  F_{m}(x) = F_{m-1}(x) + \eta \cdot h_{m}(x)
  $$
  Onde:
  - $ F_{m}(x) $ é o modelo atualizado na iteração $ m $.
  - $ F_{m-1}(x) $ é o modelo anterior.
  - $ h_{m}(x) $ é a nova árvore treinada nos resíduos.
  - $ \eta $ é a taxa de aprendizado.

### 5. Repetição do Processo

- **Iterações Múltiplas:** Repete-se o processo de calcular resíduos, treinar novas árvores e atualizar o modelo até que um número máximo de iterações seja alcançado ou até que a melhoria seja mínima.

### 6. Previsão Final

- **Modelo Composto:** A previsão final é a soma ponderada das previsões de todas as árvores construídas.


## Exemplo Ilustrativo

Imagine que queremos prever a probabilidade de um cliente cancelar sua assinatura (churn) com base em atributos como uso do serviço, satisfação e histórico de pagamentos.

### Passo a Passo:

1. **Previsão Inicial:** Começamos com uma previsão inicial, como a taxa média de churn no conjunto de dados.
2. **Cálculo dos Resíduos:** Para cada cliente, calculamos a diferença entre a previsão inicial e o valor real (se cancelou ou não).
3. **Primeira Árvore nos Resíduos:** Treinamos uma árvore para prever esses resíduos.
4. **Atualização das Previsões:** Atualizamos as previsões somando a contribuição da nova árvore, ajustada pela taxa de aprendizado.
5. **Iterações Subsequentes:** Repetimos os passos 2 a 4 várias vezes, refinando as previsões a cada iteração.
6. **Resultado Final:** A soma das previsões de todas as árvores nos dá a probabilidade final de churn para cada cliente.

## Componentes Principais do Gradient Boosting

### Função de Perda

- **Definição:** Mede o quão bem o modelo está performando (por exemplo, erro quadrático, entropia cruzada).
- **Personalização:** Pode ser escolhida de acordo com o problema (regressão, classificação binária, multiclasse).

### Taxa de Aprendizado ($ \eta $)

- **Papel:** Controla a contribuição de cada árvore para o modelo final.
- **Ajuste Fino:** Valores menores podem melhorar a generalização, mas exigem mais iterações.

### Número de Iterações (n_estimators)

- **Equilíbrio:** Um número maior de árvores pode melhorar o desempenho, mas aumenta o risco de overfitting e o tempo de treinamento.

### Profundidade das Árvores

- **Complexidade Controlada:** Árvores rasas (poucos níveis) são usadas para manter cada modelo simples e focado nos resíduos.

## Vantagens do Gradient Boosting

- **Alto Desempenho:** Pode alcançar resultados superiores em tarefas complexas.
- **Flexibilidade:** Adapta-se a diferentes tipos de dados e problemas.
- **Detecção de Padrões Complexos:** Capaz de modelar relações não lineares.

## Desvantagens do Gradient Boosting

- **Tempo de Treinamento:** Pode ser lento, especialmente com conjuntos de dados grandes.
- **Sensibilidade a Hiperparâmetros:** Requer ajuste cuidadoso para evitar overfitting.
- **Interpretação:** Mais difícil de interpretar do que modelos simples.

## Melhorias e Variações do Gradient Boosting

### XGBoost (Extreme Gradient Boosting)

- **Desempenho Otimizado:** Implementação eficiente e rápida.
- **Regularização Avançada:** Inclui termos de regularização para evitar overfitting.
- **Paralelização:** Suporta processamento paralelo para acelerar o treinamento.

### LightGBM

- **Eficiência em Grandes Dados:** Projetado para lidar com grandes volumes de dados com alta velocidade.
- **Crescimento de Árvores Leaf-wise:** Difere do crescimento nível a nível tradicional, buscando melhorar o desempenho.


# Detalhando o XGBoost em Ciência de Dados

O **XGBoost** (Extreme Gradient Boosting) é uma implementação avançada do algoritmo de **Gradient Boosting**, projetada para otimizar a velocidade e o desempenho. Desde sua introdução, tornou-se uma das ferramentas mais populares em competições de ciência de dados e aplicações do mundo real, devido à sua capacidade de produzir modelos de alta performance de forma eficiente.

## O Que é o XGBoost?

O XGBoost é uma biblioteca de software de código aberto que fornece uma implementação escalável e otimizada de algoritmos de **Gradient Boosting** para tarefas de regressão, classificação e ranking. Foi desenvolvido com foco na eficiência computacional e escalabilidade, aproveitando recursos como paralelização, regularização e tratamento eficiente de dados faltantes.

## Por Que Usar o XGBoost?

- **Alto Desempenho:** O XGBoost é conhecido por sua capacidade de produzir modelos com boas métricas.
- **Eficiência Computacional:** Otimizado para velocidade e uso de recursos, permite treinamento rápido mesmo em grandes conjuntos de dados.
- **Flexibilidade:** Suporta várias funções de perda, métricas de avaliação e opções de regularização.
- **Tratamento de Dados Faltantes:** Gerencia valores ausentes de forma inteligente durante o treinamento.
- **Paralelização e Distribuição:** Suporta execução em múltiplos núcleos e pode ser distribuído em clusters.

## Como o XGBoost Funciona?

O XGBoost aprimora o algoritmo tradicional de **Gradient Boosting** através de várias otimizações e recursos adicionais:

### 1. Regularização Avançada

- **Regularização $ L1 $ e $ L2 $:** Ajuda a evitar overfitting ao penalizar a complexidade do modelo.
- **Controle da Complexidade da Árvore:** Adiciona termos de penalização para o número de folhas e a magnitude dos pesos das folhas.

### 2. Paralelização Interna

- **Divisões Paralelas:** Realiza buscas por melhores pontos de divisão em paralelo, acelerando o treinamento.
- **Processamento em Blocos:** Lida eficientemente com dados que não cabem na memória RAM, usando processamento em blocos.

### 3. Manuseio Eficiente de Dados Faltantes

- **Divisão em Valores Faltantes:** Determina o melhor caminho (esquerda ou direita na árvore) para valores ausentes durante o treinamento.
- **Inferência Natural:** Durante a previsão, os valores faltantes seguem o mesmo caminho aprendido.

### 4. Shrinkage (Taxa de Aprendizado)

- **Atualização Gradual:** Após cada árvore, aplica-se um fator de escala (taxa de aprendizado) para reduzir o impacto de cada árvore individual, permitindo que o modelo aprenda mais lentamente e evite overfitting.

### 5. Colsample (Amostragem de Colunas)

- **Amostragem Aleatória de Atributos:** Em cada iteração, utiliza-se apenas um subconjunto dos atributos, o que aumenta a diversidade entre as árvores e reduz a correlação entre elas.

## Componentes Principais do XGBoost

### Função Objetivo Personalizável

- Permite definir funções de perda personalizadas, adequando-se a diferentes tipos de problemas.

### Funções de Avaliação

- Suporta métricas personalizadas para avaliar o desempenho durante o treinamento, como AUC, log-loss, RMSE, entre outras.

### Early Stopping (Parada Antecipada)

- Interrompe o treinamento se não houver melhoria na função de avaliação em um número definido de iterações, prevenindo overfitting e economizando tempo computacional.

## Exemplo Ilustrativo

Suponha que queremos prever o preço de casas com base em atributos como tamanho, número de quartos, localização e ano de construção.

### Passo a Passo com XGBoost:

1. **Preparação dos Dados:**
   - Carregar o conjunto de dados e lidar com valores faltantes.
   - Dividir os dados em conjuntos de treinamento e teste.

2. **Criação do Modelo:**
   - Definir os hiperparâmetros iniciais, como `n_estimators`, `max_depth`, `learning_rate`, `subsample`, `colsample_bytree`.

3. **Treinamento:**
   - O XGBoost constrói árvores sequencialmente, cada uma tentando corrigir os erros da anterior.
   - Utiliza regularização para evitar overfitting.

4. **Avaliação:**
   - Monitorar métricas como RMSE no conjunto de validação.
   - Aplicar `early_stopping_rounds` para interromper o treinamento se o modelo não melhorar.

5. **Previsão:**
   - Utilizar o modelo treinado para prever os preços no conjunto de teste.


## Vantagens do XGBoost

- **Velocidade e Eficiência:**
  - Treinamento rápido graças à paralelização e otimizações.
- **Desempenho Superior:**
  - Regularização avançada e outros recursos levam a modelos usualmente com melhores métricas.
- **Escalabilidade:**
  - Lida bem com grandes conjuntos de dados e pode ser executado em clusters.
- **Flexibilidade:**
  - Suporta uma ampla gama de cenários de aplicação e personalizações.
- **Comunidade Ativa:**
  - Extensa documentação, tutoriais e suporte da comunidade.

## Desvantagens do XGBoost

- **Complexidade de Hiperparâmetros:**
  - Muitos hiperparâmetros para ajustar, o que pode ser desafiador para iniciantes.
- **Curva de Aprendizado:**
  - Pode ser mais complexo de entender e implementar corretamente em comparação com algoritmos mais simples.
- **Recursos Computacionais:**
  - Pode exigir hardware mais potente para lidar com conjuntos de dados muito grandes.

O `score` do classificador retorna a acurácia do modelo, que é a proporção de predições corretas feitas pelo modelo. Podemos calcular a acurácia comparando as predições feitas pelo modelo com as respostas corretas. A acurácia é dada por:

$$
\text{acurácia} = \frac{\text{número de predições corretas}}{\text{número total de predições}} 
$$

Uma acurácia de 1.0 significa que o modelo fez todas as predições corretas, enquanto uma acurácia de 0.0 significa que o modelo fez todas as predições erradas.

```python
accuracy_score(y_test, y_pred)
```

### Matriz de confusão

A matriz de confusão é uma tabela que mostra o número de predições corretas e incorretas feitas pelo modelo. A matriz de confusão é dada por (considerando a ordem do Scikit-Learn):

$$
\begin{array}{|c|c|c|}
\hline
& \text{predição negativa} & \text{predição positiva} \\
\hline
\text{classe negativa} & TN & FP \\
\hline
\text{classe positiva} & FN & TP \\
\hline
\end{array}
$$


```python
ConfusionMatrixDisplay.from_predictions(y_test, y_pred, display_labels=le.classes_)
```

```python
ConfusionMatrixDisplay.from_predictions(y_test, y_pred, display_labels=le.classes_, normalize="true")
```

### Recall

Uma outra métrica que podemos usar para avaliar a qualidade de um modelo de classificação é o recall. O recall é a proporção de exemplos da classe positiva que foram corretamente classificados pelo modelo. O recall é dado por:

$$
\text{recall} = \frac{TP}{P} = \frac{TP}{TP + FN}
$$

```python
recall_score(y_test, y_pred)
```

### Precision (Precisão)

A precisão é a proporção de exemplos classificados como positivos que são realmente positivos. A precisão é dada por:

$$
\text{precisão} = \frac{TP}{\text{pred }P} = \frac{TP}{TP + FP}
$$



```python
precision_score(y_test, y_pred)
```

### F Score

O F1 score é a média harmônica entre recall e precisão. A média harmônica é usada ao invés da média aritmética porque penaliza valores extremos. O F1 score é dado por:

$$
\text{F1 score} = 2 \times \frac{\text{precisão} \times \text{recall}}{\text{precisão} + \text{recall}}
$$

Por ser uma média harmônica, o F1 score é mais sensível a valores baixos de precisão ou recall. Por exemplo, se tivermos um modelo que tem precisão 1.0 e recall 0.0, o F1 score será 0.0. Se tivermos um modelo que tem precisão 0.0 e recall 1.0, o F1 score também será 0.0.




```python
f1_score(y_test, y_pred)
```

O F1 score é um caso especial da média $F_\beta$:

$$
F_\beta = (1 + \beta^2) \times \frac{\text{precisão} \times \text{recall}}{\beta^2 \times \text{precisão} + \text{recall}}
$$

Quando $\beta = 1$, temos o F1 score. Quando $\beta > 1$, o recall tem mais peso na média. Quando $\beta < 1$, a precisão tem mais peso na média.

```python
fbeta_score(y_test, y_pred, beta=0.5)
```

```python
fbeta_score(y_test, y_pred, beta=2)
```

### Curva ROC

A curva ROC (Receiver Operating Characteristic) é uma curva que mostra a taxa de verdadeiros positivos (recall) em função da taxa de falsos positivos. A curva ROC é útil para avaliar a qualidade de um classificador binário em diferentes thresholds. A área sob a curva ROC (AUC) é uma métrica que varia entre 0.0 e 1.0, onde 0.0 significa um classificador ruim e 1.0 significa um classificador perfeito.



```python
roc_auc = roc_auc_score(y_test, y_pred)
roc_auc
```

```python
fpr, tpr, limiares = roc_curve(y_test, y_pred)

disp = RocCurveDisplay(fpr=fpr, tpr=tpr, roc_auc=roc_auc)
disp.plot(plot_chance_level=True)
```

### Curva Precision-Recall

A curva precision-recall é uma curva que mostra a precisão em função do recall. A curva precision-recall é útil para avaliar a qualidade de um classificador binário em diferentes thresholds. A área sob a curva precision-recall (AUC) é uma métrica que varia entre 0.0 e 1.0, onde 0.0 significa um classificador ruim e 1.0 significa um classificador perfeito.



```python
average_precision = average_precision_score(y_test, y_pred)
average_precision
```

```python
y_test.mean()
```

```python
precision, recall, limiares = precision_recall_curve(y_test, y_pred)

disp = PrecisionRecallDisplay(
    precision=precision,
    recall=recall,
    average_precision=average_precision,
    prevalence_pos_label=y_test.mean()
)

disp.plot(plot_chance_level=True)
```

A reta de nível de chance no gráfico de Precision-Recall representa a precisão esperada de um classificador aleatório que faz previsões positivas independentemente das características dos dados.

É a linha que indica o desempenho de um modelo que adivinha aleatoriamente as classes dos exemplos, seguindo a mesma prevalência da classe positiva no conjunto de dados. Serve como uma linha de base para comparar o desempenho do seu modelo. Se o seu modelo não superar essa linha, ele não está melhor do que uma classificação aleatória.
