# Introdução ao Gradient Boosting em Ciência de Dados

As **árvores de decisão** são algoritmos fundamentais em aprendizado de máquina, conhecidos por sua simplicidade e facilidade de interpretação. Elas funcionam dividindo os dados em subconjuntos baseados em condições sobre os atributos, criando uma estrutura em forma de árvore que leva a decisões finais. No entanto, árvores individuais podem não ser suficientemente precisas ou podem sofrer de **overfitting**. Para superar essas limitações, técnicas de ensemble como o **Gradient Boosting** são utilizadas para construir modelos mais robustos.

## Revisão: Fundamentos das Árvores de Decisão

### Como Funcionam as Árvores de Decisão?

- **Divisão Recursiva:** As árvores de decisão particionam os dados em subconjuntos menores através de condições baseadas nos atributos.
- **Estrutura em Árvore:**
  - **Nós Internos:** Representam testes ou condições sobre os atributos.
  - **Ramos:** Correspondem aos resultados dos testes, direcionando para o próximo nó.
  - **Nós Folha:** Indicam a decisão final ou previsão.
- **Critérios de Divisão:** Utilizam métricas como **Índice Gini** e **Entropia** para selecionar os melhores atributos para dividir os dados.

### Limitações das Árvores de Decisão

- **Overfitting:** Árvores muito profundas podem se ajustar excessivamente aos dados de treinamento, capturando ruídos em vez de padrões reais.
- **Instabilidade:** Pequenas variações nos dados podem levar a estruturas de árvore completamente diferentes.
- **Desempenho Subótimo:** Podem não alcançar a melhor performance possível em tarefas complexas.


## Introdução ao Gradient Boosting

### O Que é o Gradient Boosting?

O **Gradient Boosting** é um método de ensemble que constrói modelos preditivos robustos ao combinar múltiplos modelos fracos, geralmente árvores de decisão simples, de forma sequencial. Cada novo modelo é treinado para corrigir os erros cometidos pelos modelos anteriores, aprimorando progressivamente a performance geral. O termo "gradient" refere-se ao uso do gradiente da função de perda para otimizar o modelo.

### Por Que Usar o Gradient Boosting?

- **Alta Precisão:** Pode alcançar resultados superiores em comparação com modelos individuais.
- **Flexibilidade:** Suporta diversas funções de perda e pode ser aplicado a problemas de classificação e regressão.
- **Controlável:** Oferece diversos hiperparâmetros para ajustar o modelo às necessidades específicas do problema.

## Como o Gradient Boosting Funciona?

### 1. Modelo Inicial

- **Previsão Básica:** O processo começa com uma previsão inicial simples. Por exemplo, na regressão, pode ser a média dos valores alvo.

### 2. Cálculo dos Resíduos

- **Erros de Previsão:** Calcula-se a diferença entre as previsões atuais e os valores reais (resíduos).
- **Objetivo:** Os resíduos representam o que o modelo precisa aprender a prever nas próximas iterações.

### 3. Treinamento de um Novo Modelo nos Resíduos

- **Nova Árvore:** Treina-se uma nova árvore de decisão para prever os resíduos.
- **Taxa de Aprendizado ($ \eta $):** Controla o quanto cada árvore nova contribui para o modelo final. Valores menores podem levar a um modelo mais robusto.

### 4. Atualização das Previsões

- **Combinação das Previsões:** As previsões do novo modelo são combinadas com as previsões anteriores para melhorar a precisão.
- **Fórmula Geral:**
  $$
  F_{m}(x) = F_{m-1}(x) + \eta \cdot h_{m}(x)
  $$
  Onde:
  - $ F_{m}(x) $ é o modelo atualizado na iteração $ m $.
  - $ F_{m-1}(x) $ é o modelo anterior.
  - $ h_{m}(x) $ é a nova árvore treinada nos resíduos.
  - $ \eta $ é a taxa de aprendizado.

### 5. Repetição do Processo

- **Iterações Múltiplas:** Repete-se o processo de calcular resíduos, treinar novas árvores e atualizar o modelo até que um número máximo de iterações seja alcançado ou até que a melhoria seja mínima.

### 6. Previsão Final

- **Modelo Composto:** A previsão final é a soma ponderada das previsões de todas as árvores construídas.


## Exemplo Ilustrativo

Imagine que queremos prever a probabilidade de um cliente cancelar sua assinatura (churn) com base em atributos como uso do serviço, satisfação e histórico de pagamentos.

### Passo a Passo:

1. **Previsão Inicial:** Começamos com uma previsão inicial, como a taxa média de churn no conjunto de dados.
2. **Cálculo dos Resíduos:** Para cada cliente, calculamos a diferença entre a previsão inicial e o valor real (se cancelou ou não).
3. **Primeira Árvore nos Resíduos:** Treinamos uma árvore para prever esses resíduos.
4. **Atualização das Previsões:** Atualizamos as previsões somando a contribuição da nova árvore, ajustada pela taxa de aprendizado.
5. **Iterações Subsequentes:** Repetimos os passos 2 a 4 várias vezes, refinando as previsões a cada iteração.
6. **Resultado Final:** A soma das previsões de todas as árvores nos dá a probabilidade final de churn para cada cliente.

## Componentes Principais do Gradient Boosting

### Função de Perda

- **Definição:** Mede o quão bem o modelo está performando (por exemplo, erro quadrático, entropia cruzada).
- **Personalização:** Pode ser escolhida de acordo com o problema (regressão, classificação binária, multiclasse).

### Taxa de Aprendizado ($ \eta $)

- **Papel:** Controla a contribuição de cada árvore para o modelo final.
- **Ajuste Fino:** Valores menores podem melhorar a generalização, mas exigem mais iterações.

### Número de Iterações (n_estimators)

- **Equilíbrio:** Um número maior de árvores pode melhorar o desempenho, mas aumenta o risco de overfitting e o tempo de treinamento.

### Profundidade das Árvores

- **Complexidade Controlada:** Árvores rasas (poucos níveis) são usadas para manter cada modelo simples e focado nos resíduos.

## Vantagens do Gradient Boosting

- **Alto Desempenho:** Pode alcançar resultados superiores em tarefas complexas.
- **Flexibilidade:** Adapta-se a diferentes tipos de dados e problemas.
- **Detecção de Padrões Complexos:** Capaz de modelar relações não lineares.

## Desvantagens do Gradient Boosting

- **Tempo de Treinamento:** Pode ser lento, especialmente com conjuntos de dados grandes.
- **Sensibilidade a Hiperparâmetros:** Requer ajuste cuidadoso para evitar overfitting.
- **Interpretação:** Mais difícil de interpretar do que modelos simples.

## Melhorias e Variações do Gradient Boosting

### XGBoost (Extreme Gradient Boosting)

- **Desempenho Otimizado:** Implementação eficiente e rápida.
- **Regularização Avançada:** Inclui termos de regularização para evitar overfitting.
- **Paralelização:** Suporta processamento paralelo para acelerar o treinamento.

### LightGBM

- **Eficiência em Grandes Dados:** Projetado para lidar com grandes volumes de dados com alta velocidade.
- **Crescimento de Árvores Leaf-wise:** Difere do crescimento nível a nível tradicional, buscando melhorar o desempenho.


# Detalhando o LightGBM em Ciência de Dados

O **LightGBM** (Light Gradient Boosting Machine) é uma implementação de código aberto do algoritmo de **Gradient Boosting** desenvolvida pela Microsoft. Projetado para ser altamente eficiente e escalável, o LightGBM é capaz de lidar com grandes volumes de dados e oferece alta velocidade de treinamento sem sacrificar métricas. Tornou-se uma escolha popular entre os cientistas de dados devido ao seu desempenho superior em várias tarefas de aprendizado de máquina.

O LightGBM é um framework que implementa algoritmos de boosting baseados em árvores de decisão, otimizando o processo de treinamento para velocidade e uso de memória. Ele utiliza técnicas inovadoras, como o crescimento de árvores baseado em folhas e amostragem exclusiva de dados, para acelerar o treinamento e melhorar a precisão em comparação com implementações tradicionais de Gradient Boosting.

- **Alta Velocidade e Eficiência:** O LightGBM é conhecido por seu rápido tempo de treinamento, especialmente em grandes conjuntos de dados.
- **Baixo Uso de Memória:** Projetado para ser eficiente em termos de memória, permitindo lidar com grandes volumes de dados.
- **Suporte a Dados Categóricos:** Lida eficientemente com atributos categóricos sem a necessidade de extensa codificação.
- **Escalabilidade:** Capaz de ser distribuído em várias máquinas para processamento paralelo.


O LightGBM aprimora o algoritmo tradicional de **Gradient Boosting** através de várias inovações:

1. Crescimento de Árvore Leaf-wise (Baseado em Folhas)

- **Diferente do Crescimento Level-wise:** Enquanto métodos tradicionais expandem a árvore nível a nível, o LightGBM expande a folha com maior perda (ou ganho) a cada iteração.
- **Benefícios:** Este método tende a reduzir mais a função de perda e leva a um melhor ajuste dos dados.
- **Cuidado com Overfitting:** Devido ao crescimento agressivo, pode ocorrer overfitting se não houver regularização adequada.

2. Histogramas Exclusivos e Binning

- **Construção de Histogramas:** Converte os valores contínuos dos atributos em bins discretos, reduzindo o número de pontos a serem considerados.
- **Vantagens:** Reduz o custo computacional e acelera o treinamento.
- **Eficiência de Memória:** Armazena apenas os contadores dos bins, economizando memória.

3. Amostragem GOSS (Gradient-based One-Side Sampling)

- **Amostragem Inteligente:** Em vez de usar toda a amostra, o LightGBM seleciona instâncias com grandes gradientes (erros) e amostra aleatoriamente instâncias com pequenos gradientes.
- **Objetivo:** Foca em exemplos difíceis de aprender, melhorando a eficiência.

4. Exclusão de Atributos

- **Agrupamento de Atributos Mutuamente Exclusivos:** Combina atributos esparsos que raramente têm valores diferentes de zero simultaneamente.
- **Redução da Dimensionalidade:** Diminui o número de atributos a serem considerados, acelerando o treinamento.

## Componentes Principais do LightGBM

### Função Objetivo Personalizável

- **Suporte a Diversas Funções de Perda:** Como erro quadrático, log-loss, classificação lambda rank, entre outras.
- **Customização:** Permite definir funções de perda personalizadas para atender às necessidades específicas do problema.

### Tratamento Eficiente de Atributos Categóricos

- **Sem Necessidade de One-Hot Encoding:** Lida diretamente com atributos categóricos, preservando a informação de ordenação e hierarquia.
- **Algoritmo Otimizado:** Utiliza técnicas específicas para dividir atributos categóricos de forma eficiente.

### Paralelismo e Distribuição

- **Execução Paralela:** Suporta treinamento paralelo em múltiplos núcleos de CPU.
- **Treinamento Distribuído:** Pode ser executado em clusters, aproveitando recursos de várias máquinas.



## Vantagens do LightGBM

- **Alta Velocidade de Treinamento:** Graças a otimizações como histogramas exclusivos e crescimento leaf-wise.
- **Baixo Consumo de Memória:** Uso eficiente de memória, permitindo lidar com grandes conjuntos de dados.
- **Suporte a Atributos Categóricos:** Tratamento eficiente sem necessidade de pré-processamento complexo.
- **Alta Precisão:** Melhoria da performance através de técnicas avançadas.
- **Escalabilidade:** Capacidade de treinar modelos em ambientes distribuídos.

## Desvantagens do LightGBM

- **Risco de Overfitting:** O crescimento leaf-wise pode levar a overfitting se não houver regularização adequada.
- **Sensibilidade a Hiperparâmetros:** Requer ajuste cuidadoso dos hiperparâmetros para alcançar o melhor desempenho.
- **Curva de Aprendizado:** Pode ser mais complexo de configurar corretamente em comparação com algoritmos mais simples.



# Comparação entre XGBoost e LightGBM

**XGBoost** e **LightGBM** são dois dos algoritmos de **Gradient Boosting** mais populares em ciência de dados. Ambos são amplamente utilizados devido ao seu alto desempenho e eficiência em tarefas de aprendizado de máquina, como classificação e regressão. Apesar de suas semelhanças, existem diferenças fundamentais em seus aspectos técnicos, interpretabilidade e outros fatores que podem influenciar a escolha de um sobre o outro em diferentes situações.

## Aspectos Fundamentais

### 1. Estratégias de Crescimento de Árvore

#### XGBoost

- **Crescimento Level-wise (Nível a Nível):**
  - As árvores são construídas adicionando nós em todos os níveis de forma equilibrada.
  - **Vantagens:**
    - Tende a reduzir o overfitting, pois o crescimento é mais controlado.
    - A construção equilibrada das árvores pode facilitar a interpretação.
  - **Desvantagens:**
    - Pode ser mais lento em comparação com estratégias mais agressivas.

#### LightGBM

- **Crescimento Leaf-wise (Baseado em Folhas):**
  - Expande a folha com maior perda (ou ganho) a cada iteração, independentemente do nível.
  - **Vantagens:**
    - Geralmente alcança menor perda com o mesmo número de folhas.
    - Acelera o processo de treinamento.
  - **Desvantagens:**
    - Maior risco de overfitting, especialmente com dados menores.
    - Pode resultar em árvores desequilibradas, tornando a interpretação mais complexa.

### 2. Tratamento de Atributos Categóricos

#### XGBoost

- **Codificação Necessária:**
  - Requer que atributos categóricos sejam convertidos em numéricos, geralmente através de **One-Hot Encoding** ou **Label Encoding**.
  - **Implicações:**
    - Aumento da dimensionalidade com One-Hot Encoding, o que pode afetar o desempenho.
    - Potencial perda de informação sobre a hierarquia ou ordens naturais nas categorias.

#### LightGBM

- **Suporte Nativo a Atributos Categóricos:**
  - Lida diretamente com atributos categóricos sem necessidade de codificação explícita.
  - Utiliza técnicas como **Gradient-based One-Side Sampling (GOSS)** para otimizar a divisão em atributos categóricos.
  - **Vantagens:**
    - Mantém a integridade dos dados categóricos.
    - Reduz o pré-processamento e potencialmente melhora a performance.

### 3. Otimizações e Técnicas Avançadas

#### XGBoost

- **Regularização Avançada:**
  - Suporta regularização $ L1 $ e $ L2 $, ajudando a evitar overfitting.
- **Paralelização Interna:**
  - Realiza paralelização nas buscas por melhores pontos de divisão.
- **Suporte a Árvores Lineares:**
  - Permite o uso de modelos lineares como aprendizes base.

#### LightGBM

- **Histogramas Exclusivos e Binning:**
  - Constrói histogramas para acelerar o treinamento e reduzir o uso de memória.
- **Amostragem GOSS:**
  - Foca em instâncias com grandes gradientes (erros), aumentando a eficiência.
- **Exclusão de Atributos:**
  - Combina atributos mutuamente exclusivos para reduzir a dimensionalidade.

### 4. Desempenho e Eficiência

#### Velocidade de Treinamento

- **XGBoost:**
  - Rápido e eficiente, mas geralmente mais lento que o LightGBM em conjuntos de dados muito grandes.
- **LightGBM:**
  - Projetado para velocidade, geralmente mais rápido que o XGBoost devido a suas otimizações.

#### Uso de Memória

- **XGBoost:**
  - Eficiente, mas pode consumir mais memória em comparação com o LightGBM.
- **LightGBM:**
  - Usa menos memória graças ao binning e outras otimizações.

#### Escalabilidade

- **XGBoost:**
  - Suporta paralelização e pode ser executado em clusters.
- **LightGBM:**
  - Altamente escalável, com suporte robusto para treinamento distribuído em grandes volumes de dados.

## Interpretabilidade

### Complexidade do Modelo

- **XGBoost:**
  - As árvores level-wise tendem a ser mais equilibradas, o que pode facilitar a interpretação visual.
- **LightGBM:**
  - As árvores leaf-wise podem ser profundas e desequilibradas, tornando a interpretação mais desafiadora.

### Importância dos Atributos

Ambos os algoritmos fornecem medidas de importância dos atributos, permitindo que os usuários entendam quais variáveis estão influenciando mais o modelo.

- **XGBoost:**
  - Fornece métricas como ganho, cobertura e frequência para avaliar a importância dos atributos.
- **LightGBM:**
  - Fornece medidas similares, mas devido ao crescimento leaf-wise, a distribuição de importância pode ser diferente.

## Outros Aspectos Importantes

### Tratamento de Valores Faltantes

- **XGBoost:**
  - Trata valores faltantes determinando o melhor caminho (esquerda ou direita) para seguir durante o treinamento.
- **LightGBM:**
  - Lida automaticamente com valores faltantes, tratando-os como uma categoria separada ou seguindo estratégias similares ao XGBoost.

### Hiperparâmetros e Ajuste

- **Complexidade de Hiperparâmetros:**
  - Ambos possuem muitos hiperparâmetros que podem ser ajustados para otimizar o desempenho.
- **Sensibilidade:**
  - O LightGBM pode ser mais sensível aos hiperparâmetros devido ao seu crescimento leaf-wise, exigindo ajuste cuidadoso para evitar overfitting.

### Robustez e Overfitting

- **XGBoost:**
  - Com suas técnicas de regularização e crescimento controlado de árvores, tende a ser mais robusto contra overfitting.
- **LightGBM:**
  - Pode overfitar mais facilmente se não houver regularização adequada, devido ao seu crescimento agressivo de árvores.


## Resumo Comparativo

## Considerações Finais

A escolha entre **XGBoost** e **LightGBM** depende de vários fatores, incluindo o tamanho e a natureza dos dados, a necessidade de interpretabilidade, os recursos computacionais disponíveis e os objetivos específicos do projeto.

- **Use XGBoost se:**
  - Precisa de maior controle para evitar overfitting.
  - A interpretabilidade do modelo é uma prioridade.
  - Está trabalhando com conjuntos de dados menores ou médios.

- **Use LightGBM se:**
  - Está lidando com conjuntos de dados muito grandes e precisa de alta velocidade.
  - Deseja suporte nativo a atributos categóricos sem pré-processamento adicional.
  - Tem recursos computacionais limitados e precisa de eficiência de memória.
  - Está disposto a investir tempo no ajuste cuidadoso de hiperparâmetros para evitar overfitting.

$$
\text{acurácia} = \frac{\text{número de predições corretas}}{\text{número total de predições}} 
$$

Uma acurácia de 1.0 significa que o modelo fez todas as predições corretas, enquanto uma acurácia de 0.0 significa que o modelo fez todas as predições erradas.

### Matriz de confusão

A matriz de confusão é uma tabela que mostra o número de predições corretas e incorretas feitas pelo modelo. A matriz de confusão é dada por (considerando a ordem do Scikit-Learn):

$$
\begin{array}{|c|c|c|}
\hline
& \text{predição negativa} & \text{predição positiva} \\
\hline
\text{classe negativa} & TN & FP \\
\hline
\text{classe positiva} & FN & TP \\
\hline
\end{array}
$$

### Recall

Uma outra métrica que podemos usar para avaliar a qualidade de um modelo de classificação é o recall. O recall é a proporção de exemplos da classe positiva que foram corretamente classificados pelo modelo. O recall é dado por:

$$
\text{recall} = \frac{TP}{P} = \frac{TP}{TP + FN}
$$

```python
recall_score(y_test, y_pred)
```

### Precision (Precisão)

A precisão é a proporção de exemplos classificados como positivos que são realmente positivos. A precisão é dada por:

$$
\text{precisão} = \frac{TP}{\text{pred }P} = \frac{TP}{TP + FP}
$$



```python
precision_score(y_test, y_pred)
```

### F Score

O F1 score é a média harmônica entre recall e precisão. A média harmônica é usada ao invés da média aritmética porque penaliza valores extremos. O F1 score é dado por:

$$
\text{F1 score} = 2 \times \frac{\text{precisão} \times \text{recall}}{\text{precisão} + \text{recall}}
$$

Por ser uma média harmônica, o F1 score é mais sensível a valores baixos de precisão ou recall. Por exemplo, se tivermos um modelo que tem precisão 1.0 e recall 0.0, o F1 score será 0.0. Se tivermos um modelo que tem precisão 0.0 e recall 1.0, o F1 score também será 0.0.




```python
f1_score(y_test, y_pred)
```

O F1 score é um caso especial da média $F_\beta$:

$$
F_\beta = (1 + \beta^2) \times \frac{\text{precisão} \times \text{recall}}{\beta^2 \times \text{precisão} + \text{recall}}
$$

Quando $\beta = 1$, temos o F1 score. Quando $\beta > 1$, o recall tem mais peso na média. Quando $\beta < 1$, a precisão tem mais peso na média.

```python
fbeta_score(y_test, y_pred, beta=0.5)
```

```python
fbeta_score(y_test, y_pred, beta=2)
```

### Curva ROC

A curva ROC (Receiver Operating Characteristic) é uma curva que mostra a taxa de verdadeiros positivos (recall) em função da taxa de falsos positivos. A curva ROC é útil para avaliar a qualidade de um classificador binário em diferentes thresholds. A área sob a curva ROC (AUC) é uma métrica que varia entre 0.0 e 1.0, onde 0.0 significa um classificador ruim e 1.0 significa um classificador perfeito.



```python
roc_auc = roc_auc_score(y_test, y_pred)
roc_auc
```

### Curva Precision-Recall

A curva precision-recall é uma curva que mostra a precisão em função do recall. A curva precision-recall é útil para avaliar a qualidade de um classificador binário em diferentes thresholds. A área sob a curva precision-recall (AUC) é uma métrica que varia entre 0.0 e 1.0, onde 0.0 significa um classificador ruim e 1.0 significa um classificador perfeito.



```python
average_precision = average_precision_score(y_test, y_pred)
average_precision
```

A reta de nível de chance no gráfico de Precision-Recall representa a precisão esperada de um classificador aleatório que faz previsões positivas independentemente das características dos dados.

É a linha que indica o desempenho de um modelo que adivinha aleatoriamente as classes dos exemplos, seguindo a mesma prevalência da classe positiva no conjunto de dados. Serve como uma linha de base para comparar o desempenho do seu modelo. Se o seu modelo não superar essa linha, ele não está melhor do que uma classificação aleatória.
